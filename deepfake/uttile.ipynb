```python
import tensorflow as tf
from tensorflow import keras
from keras import Sequential
from keras.layers import Dense, Conv2D, MaxPooling2D, Flatten,BatchNormalization,Dropout
```

```python
# generators    # generators helps lo load the data batch wise becous thousent of images not directly load its devide batchwise and execute batch by batch
train_ds = keras.utils.image_dataset_from_directory(
    directory=r'C:\Users\Gaurav Ghugarkar\OneDrive\Desktop\Data science\DL\CNN\AI_image_detector\dataset',
    labels='inferred',
    label_mode='int',
    batch_size=32,                          # lable mode means automatic provide lable 0,1 cat,dog
    image_size=(256, 256)
)

# its return (32, 256, 256, 3) 32->images  per batch  shape
```

```python
# Normalize
def process(image, label):
    image = tf.cast(image / 255., tf.float32)
    return image, label

train_ds = train_ds.map(process)
```

```python
import tensorflow
from tensorflow import keras
from keras import Sequential
from keras.layers import Dense,Flatten
from keras.applications.vgg16 import VGG16
```

```python
conv_base = VGG16(
    weights='imagenet',
    include_top = False,
    input_shape=(256,256,3)
)
```

```python
conv_base.summary()
```

```python
conv_base.trainable = False  #  means freaze all cnn layers  only training appay on ann layers
```

```python
model = Sequential()

model.add(conv_base)
model.add(Flatten())
model.add(Dense(256,activation='relu'))
model.add(Dense(1,activation='sigmoid'))
```

```python
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
history=model.fit(train_ds, epochs=10)     #for x, y in train_ds:
                                                                          # print(x.shape)  # images
                                                                          # print(y.shape)  # labels
                                                                          # break
```
